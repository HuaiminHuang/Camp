# 栈
栈（stack）是一种遵循**先入后出**逻辑的线性数据结构。
![alt text](stack_operations.png)

```python
# 初始化栈
# Python 没有内置的栈类，可以把 list 当作栈来使用
stack: list[int] = []

# 元素入栈
stack.append(1)
stack.append(3)
stack.append(2)
stack.append(5)
stack.append(4)

# 访问栈顶元素
peek: int = stack[-1]
print(peek, stack)

# 元素出栈
pop: int = stack.pop()
print(pop, stack)

# 获取栈的长度
size: int = len(stack)
print(size)

# 判断是否为空
is_empty: bool = len(stack) == 0
print(is_empty)
```

基于链表的栈构建

```python
class ListNode:
    """链表节点类"""
    def __init__(self, val):
        self.val = val
        self.next = ListNode | None


class LinkedListStack:
    """基于链表实现的栈"""

    def __init__(self):
        """构造方法"""
        self._peek: ListNode | None = None  #### 存一个peek以便访问顶端元素
        self._size: int = 0

    def size(self) -> int:
        """获取栈的长度"""
        return self._size

    def is_empty(self) -> bool:
        """判断栈是否为空"""
        return self._size == 0

    def push(self, val: int):
        """入栈"""
        node = ListNode(val)
        node.next = self._peek
        self._peek = node
        self._size += 1

    def pop(self) -> int:
        """出栈"""
        num = self.peek()
        self._peek = self._peek.next
        self._size -= 1
        return num

    def peek(self) -> int:
        """访问栈顶元素"""
        if self.is_empty():
            raise IndexError("栈为空")
        return self._peek.val

    def to_list(self) -> list[int]:
        """转化为列表用于打印"""
        arr = []
        node = self._peek
        while node:
            arr.append(node.val)
            node = node.next
        arr.reverse()
        return arr
    
    # def __repr__(self) -> str:
    #     """返回栈的字符串表示"""
    #     arr = []
    #     node = self._peek
    #     while node:
    #         arr.append(node.val)
    #         node = node.next
    #     arr.reverse()
    #     return f"LinkedListStack({arr})"
    
if __name__ == "__main__":
    # 初始化
    stack = LinkedListStack()

    # 元素入栈
    stack.push(1)
    stack.push(3)
    stack.push(2)
    stack.push(5)
    stack.push(4)
    print(stack, stack.to_list())

    # 访问栈顶元素
    peek = stack.peek()
    print("栈顶元素 peek =", peek)

    # 元素出栈
    pop = stack.pop()
    print("出栈的元素：", pop,"\n现在的stack: ", stack.to_list())
    pop = stack.pop()
    print("出栈的元素：", pop,"\n现在的stack: ", stack.to_list())

    # 获取长度
    size = stack.size()
    print("size为：", size)
    
    # 判断非空
    is_empty = stack.is_empty()
    print("栈是否非空：", is_empty)
```

基于(动态)数组的栈构建

```python
class ListNode:
    """链表节点类"""
    def __init__(self, val):
        self.val = val
        self.next = ListNode | None

class ArrayStack:
    """基于数组实现的栈"""

    def __init__(self):
        """构造方法"""
        self._stack: list[int] = []

    def size(self) -> int:
        """获取栈的长度"""
        return len(self._stack)

    def is_empty(self) -> bool:
        """判断栈是否为空"""
        return self.size() == 0

    def push(self, item: int):
        """入栈"""
        self._stack.append(item)

    def pop(self) -> int:
        """出栈"""
        if self.is_empty():
            raise IndexError("栈为空")
        return self._stack.pop()

    def peek(self) -> int:
        """访问栈顶元素"""
        if self.is_empty():
            raise IndexError("栈为空")
        return self._stack[-1]

    def to_list(self) -> list[int]:
        """返回列表用于打印"""
        return self._stack
    
if __name__ == "__main__":
    # 初始化
    stack = LinkedListStack()

    # 元素入栈
    stack.push(5)
    stack.push(1)
    stack.push(2)
    stack.push(5)
    stack.push(8)
    print(stack, stack.to_list())

    # 访问栈顶元素
    peek = stack.peek()
    print("栈顶元素 peek =", peek)

    # 元素出栈
    pop = stack.pop()
    print("出栈的元素：", pop,"\n现在的stack: ", stack.to_list())
    pop = stack.pop()
    print("出栈的元素：", pop,"\n现在的stack: ", stack.to_list())

    # 获取长度
    size = stack.size()
    print("size为：", size)
    
    # 判断非空
    is_empty = stack.is_empty()
    print("栈是否非空：", is_empty)
```

# 队列（queue）
队列（queue）是一种遵循先入先出规则的线性数据结构。顾名思义，队列模拟了排队现象，即新来的人不断加入队列尾部，而位于队列头部的人逐个离开。

为了实现队列，我们需要一种数据结构，可以在**一端添加元素，并在另一端删除元素**，链表和数组都符合要求。

1. linked list (链表) 的队列实现

```python
class ListNode:
    def __init__(self, val):
        self.val = val
        self.next:  ListNode | None = None

class LinkedQueue:
    """基于链表的实现"""
    def __init__(self):
        self._front: ListNode | None = None # 头节点
        self._rear: ListNode | None = None # 尾节点
        self._size: int = 0

    def size(self) -> int:
        """获取size"""
        return self._size
    
    def is_empty(self) -> bool:
        """判断是否非空"""
        return self._size == 0
        
    def push(self, num: int):
        """入队"""
        # 在尾节点添加 num
        node = ListNode(num)
        # 如果列队empty，头尾都指向该节点
        if self._front is None:
            self._front = node
            self._rear = node
        else:
            self._rear.next = node
            self._rear = node
        self._size += 1

    def pop(self) -> int:
        """出队并返回头值"""
        num = self.peek()
        self._front = self._front.next
        self._size -= 1
        return num


    def peek(self) -> int:
        if self.is_empty():
            raise IndexError("列表为空")
        return self._front.val


    def to_list(self) -> list[int]:
        """转化为列表方便打印"""
        queue = []
        temp = self._front
        while temp:
            queue.append(temp.val)
            temp = temp.next
        return queue
    

if __name__ =="__main__":
    que = LinkedQueue()

    # 入队
    que.push(12)
    print(f"queue: {que.to_list()}, 队头/队尾： {que._front.val}/{que._rear.val}")
    que.push(19)
    que.push(4)
    que.push(11)
    que.push(5)
    que.push(92)
    print(f"queue: {que.to_list()}, 队头/队尾： {que._front.val}/{que._rear.val}")

    # 访问队首
    peek = que.peek()
    print("队首元素：", peek, f"队头/队尾： {que._front.val}/{que._rear.val}")

    # 元素出队
    pop = que.pop()
    print("出队元素：", pop, f"队头/队尾： {que._front.val}/{que._rear.val}")

    # 获取尺寸和是否为空
    print(f"size is: {que.size()}, is empty: {que.is_empty()}" )
```

2. 基于数组的队列实现
数组的删除首元素的时间复杂度是 $O(n)$，可以通过变量 `front` 指向队首元素的索引，并维护一个变量 `size` 用于记录队列长度。`rear = front + size` ，这个公式计算出的 rear 指向队尾元素之后的下一个位置。
![alt text](array_queue_step2_push.png)

```python
class ArrayQueue:
    """基于环形数组实现的队列"""

    def __init__(self, size: int):
        """构造方法"""
        self._nums: list[int] = [0] * size  # 用于存储队列元素的数组
        self._front: int = 0  # 队首指针，指向队首元素
        self._size: int = 0  # 队列长度

    def capacity(self) -> int:
        """获取队列的容量"""
        return len(self._nums)

    def size(self) -> int:
        """获取队列的长度"""
        return self._size

    def is_empty(self) -> bool:
        """判断队列是否为空"""
        return self._size == 0

    def push(self, num: int):
        """入队"""
        if self._size == self.capacity():
            raise IndexError("队列已满")
        # 计算队尾指针，指向队尾索引 + 1
        # 通过取余操作实现 rear 越过数组尾部后回到头部
        rear: int = (self._front + self._size) % self.capacity()
        # 将 num 添加至队尾
        self._nums[rear] = num
        self._size += 1

    def pop(self) -> int:
        """出队"""
        num: int = self.peek()
        # 队首指针向后移动一位，若越过尾部，则返回到数组头部
        self._front = (self._front + 1) % self.capacity()
        self._size -= 1
        return num

    def peek(self) -> int:
        """访问队首元素"""
        if self.is_empty():
            raise IndexError("队列为空")
        return self._nums[self._front]

    def to_list(self) -> list[int]:
        """返回列表用于打印"""
        res = [0] * self.size()
        j: int = self._front
        for i in range(self.size()):
            res[i] = self._nums[(j % self.capacity())]
            j += 1
        return res
    
if __name__ =="__main__":
    que = ArrayQueue(7)

    # 入队
    que.push(12)
    print(f"queue: {que.to_list()}")
    que.push(19)
    que.push(4)
    que.push(11)
    que.push(5)
    que.push(92)
    print(f"queue: {que.to_list()}")

    # 访问队首
    peek = que.peek()
    print("队首元素：", peek)

    # 元素出队
    pop = que.pop()
    print("出队元素：", pop)
    peek = que.peek()
    print("队首元素：", peek)

    # 获取尺寸和是否为空
    print(f"size is: {que.size()}, is empty: {que.is_empty()}" )
```
