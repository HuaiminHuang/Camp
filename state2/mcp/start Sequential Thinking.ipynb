# mcp服务本地运行


- 环境：
    - 客户端[ChatBox](https://chatboxai.app/zh)
    
    - 测试对象[Sequential Thinking](https://github.com/modelcontextprotocol/servers/tree/main/src/sequentialthinking)

    - conda进行环境隔离

## 1. 创建激活环境
- 创建新的 conda 环境（推荐使用 Python 3.9+, nodejs=20 (这里st需要的版本)）

```bash
conda create -n mcp-st python=3.10 nodejs=20
conda activate mcp-st
```

- 检查环境

```bash
node -v
npm -v 
```

## 2. 克隆并且进入servers
```bash
git clone https://github.com/modelcontextprotocol/servers.git
cd servers/src/sequentialthinking
```
或者也可直接从代码仓库进行下载解压到指定目录位置

```bash
# 清理可能的旧安装
rm -rf node_modules
rm -f package-lock.json

# 安装依赖
npm install
```

## 3. 构建项目

```bash
# 构建 TypeScript 项目
npm run build

# 或者直接运行（如果 package.json 中有开发脚本）
npm run dev
```

## 4. 运行服务器

```bash
# 运行服务器
npm start

# 或者直接运行构建后的文件
node dist/index.js

# 或者
npx mcp-server-sequential-thinking
```

## 常见问题解决

安装前请先阅读README.md

### 1. chatbox 找不到 npx

需要 `控制面板-系统-高级设置-环境变量`

在系统环境变量的path创建conda淡妆npx所在的位置

一般来说是在：`C:\Users\hhm18\miniconda3\envs\mcp `

### 2. 对于使用pip安装的可以参考

```md
{
  "mcpServers": {
    "fetch": {
      "command": "C:\\Users\\hhm18\\miniconda3\\envs\\mcp\\`python.exe`",
      "args": ["-m", "mcp_server_fetch"]
    }
  }
}
```
