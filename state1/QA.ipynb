```python
from torch.utils.data import Dataset
import json

class DuReaderQG(Dataset):
    def __init__(self, data_file):
        self.data = self.load_data(data_file)

    def load_data(self,data_file):
        Data = {}
        with open(data_file, 'rt', encoding='utf-8') as f:
            for idx, line in enumerate(f):
                sample = json.loads(line.strip())
                Data[idx] = sample
        return Data
    
    def __len__(self):
        return len(self.data)

    def __getitem__(self, idx):
        return self.data[idx]
```

```python
import sacrebleu

references = [
    ["这 是 一个 测试 句子", "这 是 一个 示例 文本"]
]
hypotheses = [
    "这 是 一个 测试 文本",
    "这 是 一个 示例 句子"
]

# 计算BLEU
bleu = sacrebleu.corpus_bleu(hypotheses, references)

# 总体BLEU分数
print("BLEU score:", bleu.score)

# 各 n-gram 的精确度
print("BLEU-1~4 precisions:", bleu.precisions)  # 返回列表 [p1, p2, p3, p4]

# 其他指标
print("BP:", bleu.bp)
print("ratio:", bleu.sys_len / bleu.ref_len)
print("hyp_len:", bleu.sys_len)
print("ref_len:", bleu.ref_len)
```

```python
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM, T5Model

tokenizer = AutoTokenizer.from_pretrained("Langboat/mengzi-T5-base")
model = AutoModelForSeq2SeqLM.from_pretrained("Langboat/mengzi-T5-base")

message = "对话：你是谁？"
inputs = tokenizer(message, return_tensors="pt")

output = model.generate(
    **inputs,
    max_length=50,
    do_sample=True,
    top_k=20,
    top_p=0.8,
    temperature=0.7,
    num_return_sequences=1,
    pad_token_id=tokenizer.eos_token_id  # 明确指定pad token
)

print(tokenizer.decode(output[0], skip_special_tokens=True))
```

```python
weights_list = [
    (1., 0, 0, 0),               # BLEU-1
    (0.5, 0.5, 0, 0),            # BLEU-2
    (1/3, 1/3, 1/3, 0),          # BLEU-3
    (0.25, 0.25, 0.25, 0.25)     # BLEU-4
]

scores = [sentence_bleu(reference, candidate, weights=w) for w in weights_list]
print(scores)  # [BLEU1, BLEU2, BLEU3, BLEU4]
```
